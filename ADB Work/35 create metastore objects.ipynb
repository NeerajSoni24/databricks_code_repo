Create Database within the catalog

```python
%sql
create database if not exists dev.demo_db;
```

```python
%sql
GRANT USE CATALOG ON CATALOG dev TO `adb6pm-dev`;
GRANT ALL PRIVILEGES ON DATABASE dev.demo_db TO `adb6pm-dev`;
```
