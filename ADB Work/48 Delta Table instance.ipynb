Delta Table instance:

  -To make replica to actual delta table

Advantage of instance: To perform sql operations using pyspark

syntax: DeltaTable.forName(spark,"tablename")

```python
#step 1: Creating delta table

from delta.tables import *
#DeltaTable.create(spark) \
DeltaTable.createIfNotExists(spark)\
     .tableName("dev.demo_db.emp3") \
     .addColumn("eid","INT") \
     .addColumn("ename","STRING") \
     .addColumn("sal","DOUBLE") \
     .addColumn("sex","STRING") \
    .addColumn("dno","INT") \
    .execute()
#Delta table is now created
```

step 2: inserting data into Delta table

```python

%sql
insert into dev.demo_db.emp3 values 
(101,"Raj",50000,"M",11),
(102,"Ravi",60000,"M",12),
(103,"Suresh",70000,"M",13);
```

step 3: Retrieving data

```python
%sql
select * from dev.demo_db.emp3;
```

Step 4: Creating instance for delta table

```python
deltainstance_emp3=DeltaTable.forName(spark,"dev.demo_db.emp3")
display(deltainstance_emp3.toDF())
```

Step 5: Now perform operations both on table and instance

Table: dev.demo_db.emp3

Instance: deltainstance_emp3

```python
%sql
 describe dev.demo_db.emp3;
```

```python
#using pyspark-->describe the instance
display(deltainstance_emp3.toDF().printSchema())
```

Describe the history of both tables


```python
%sql
 describe  history dev.demo_db.emp3;
```

using pyspark --->describing history

```python
display(deltainstance_emp3.history())
```

a. using sql
