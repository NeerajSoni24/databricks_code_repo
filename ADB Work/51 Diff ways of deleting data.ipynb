```python
# 2nd method : spark sql

spark.sql("delete from dev.demo_db.emp4 where eid=106 ")
```

```python
%sql
select * from dev.demo_db.emp4;
```

Now EID 106 also deleted.

```python
# Select data from spark sql.
spark.sql("select * from dev.demo_db.emp4 ").display()
```

```python
# 3rd method deletion : Pyspark Delta Table instance

from delta.tables import *
from pyspark.sql.functions import *

deltatable_instance=DeltaTable.forName(spark,"dev.demo_db.emp4")
deltatable_instance.delete("eid=105")
```

Retrieve the data

EID 105 should be deleted from the table

```python
%sql
select * from dev.demo_db.emp4;
```

```python
# ex:2 multiple conditions using sql predicate

deltatable_instance.delete("eid=104 and ename='Rajesh'")
```

```python
%sql
select * from dev.demo_db.emp4;
```
