Window Functions in pyspark:

The available window functions are:

1) Rownumber

2) rank

2) denserank


On which column you want to apply window function---> sal column

```python
df=spark.read.csv("/mnt/files/csvdata/emp.csv",header=True,inferSchema=True)
df.show()
```

```python
#First import the Window function and then sort by sal
from pyspark.sql.window import Window

window=Window.orderBy(df.Salary.desc())

print(window)
```

o/p: <class 'pyspark.sql.window.Window'>

1) row_number(): 
- assigning a number to each row
- number will be unique and auto-incremented

Now applying row_number over the window

whatever col is present in the window for that this row_number is applied.

```python
from pyspark.sql.functions import row_number, rank, dense_rank

df_row=df.withColumn("Row_Number",row_number().over(window))
df_row.show()

df_rank=df.withColumn("rank",rank().over(window))
df_rank.show()

df_denserank=df.withColumn("dense_rank",dense_rank().over(window))
df_denserank.show()
```

here we are getting same-->for rownumber,rank and denserank
because there are no duplicates.

1) row_numer(): assigning a number to each row
                number will be unique and auto-incremented


```python
print("Total count of the rows")
count=df_row.count()
print(count)
```

```python
from pyspark.sql.functions import min,max

print("Finding min Row_Number:")
df_row.agg(min("Row_Number")).show()

print("Finding max Row_Number")
df_row.agg(max("Row_Number")).show()
```

2) Rank() function:
- assigning rank allocated to each row within a group of datasets
- here some ranks will be skipped, if there are duplicates

ex: sal

   Rowno    sal          Rank

     1    10000     1

     2    10000     1

     3    10000     1

     4    20000     4 after 3ranks-->1,1,1-->next rank is->it takes the rowno 4 as rank number.
     
     5    20000     4
     
     6    20000     4
     
     7    30000     7 after 3ranks-->4,4,4-->next rank is->it takes the rowno 7 as rank number.


3) Dense Rank function:

dense Rank--->
- assigning rank for each group of values
- here some ranks wont be skipped, if there are duplicates

ex: sal

   Rowno  sal      Rank

     1    10000     1

     2    10000     1

     3    10000     1

     4    20000     2  

     5    20000     2

     6    20000     2

     7    30000     3 

dense rank 
ex:2

- If 1st 10 emps(or 10 rows) has salary--->10000->here 1st 10 rows will get Rank1
- next   10 emps(or 10 rows) has salary--->20000->    next 10 rows will get Rank2

note: in case of rank function-->1st 10 rows will get rank1 

and next 10 rows will get rank 11.


Rank  and dense_rank function:

```python
df=spark.read.csv("/mnt/files/csvdata/emp7.csv",header=True,inferSchema=True)
df.show()
```

```python
from pyspark.sql.functions import rank,dense_rank,row_number
from pyspark.sql.window import Window

window=Window.orderBy(df.Salary.desc()) 

df_row=df.withColumn("Row_Number",row_number().over(window))
display(df_row)
```

```python
df_rank=df_row.withColumn("rank",rank().over(window))
display(df_rank)
```

```python
df_denserank=df_row.withColumn("dense_rank",dense_rank().over(window))
display(df_denserank)
```

Relatieron b/w rank and Row_number----> always 

Rank <= Row_number

Now create a df and get the 3 columns----> rowno, rank, dense_rank

```python
df_denserank1=df_rank.withColumn("dense_rank",dense_rank().over(window))
display(df_denserank1)
```

just display - eid, ename, sal, rownumber,rank and denserank 

for those whose sal>=30000

```python
df_res=df_denserank1.filter(df_denserank1['Salary']>=30000)
display(df_res)

display(df_res.select(["eid","ename","Salary","Row_Number","rank","dense_rank"]))
```
