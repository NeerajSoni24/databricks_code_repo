Task1: ADB reading data from Azure sql table(emp) and writing into ADLS

```python
jdbcHostname="mydbserver2021.database.windows.net"
jdbcPort=1433
jdbcDatabase="azsqldb2021"
jdbcUsername="mydbuser2021"
jdbcPassword="Neeraj@2021"
jdbcDriver="com.microsoft.sqlserver.jdbc.SQLServerDriver"

jdbcUrl=f"jdbc:sqlserver://{jdbcHostname}:{jdbcPort};databaseName={jdbcDatabase};user={jdbcUsername};password={jdbcPassword}"
```

```python
df1=spark.read.format("jdbc").option("url",jdbcUrl).option("dbtable", "dbo.emp").load()
display(df1)
```

Writing Dataframe into ADLS

- It creates a new folder under the mount point (/mnt/files/...)

- Here /mnt/files is a mount point which is referring to 'dbfscontainer' in ADLS 'mystoragelakeadb6pmgroup' datalake.

```python
df1.write.format("csv").option("header", "true").mode("overwrite").save("/mnt/files/empres3")
```
