Different ways on inserting data into delta table

Ist method : SQL Style insert



```python
# Step 1: Create Delta Table
from delta.tables import *

DeltaTable.create(spark) \
     .tableName("dev.demo_db.emp4") \
     .addColumn("eid","INT") \
     .addColumn("ename","STRING") \
     .addColumn("sal","DOUBLE") \
     .addColumn("sex","STRING") \
    .addColumn("dno","INT") \
    .execute()
```

Step 2: inserting data(sql style)

```python
%sql
insert into dev.demo_db.emp4 values (101,"Raj",50000,"M",11),(102,"Ravi",60000,"M",12),(103,"Suresh",70000,"M",13);
```

Step 3: Retrieving data

SQL style

```python
%sql
select * from dev.demo_db.emp4;
```

pyspark style 

show, display data

```python
spark.sql("select * from dev.demo_db.emp4").show()

# another way to see data in table format
spark.sql("select * from dev.demo_db.emp4").display()

# another way to see data in table format
display(spark.sql("select * from dev.demo_db.emp4"))
```

```python
#Step 1: creating or defining schema-->multiple ways

#I-way
schema1="""eid int,
           ename string,
           sal double,
           sex string,
           dno int"""
#or

#II-way
#from pyspark.sql.types import *
#Or
from pyspark.sql.types import StructType,StructField,IntegerType,StringType,DoubleType
schema2=StructType([
         StructField("eid",IntegerType(),False), \
         StructField("ename",StringType(),True), \
         StructField("sal",DoubleType(),True), \
         StructField("sex",StringType(),True), \
         StructField("dno",IntegerType(),True) \
         ])

#here True--->means nullable is True
```

```python
#Step 2: Defining data
data1=[(104,"Rajesh",40000.00,"M",13),(105,"Ramesh",50000.00,"M",12)]

#step 3:creating dataframe
df1=spark.createDataFrame(data=data1,schema=schema2)
#or
#df1=spark.createDataFrame(data1,schema2)
```

```python
#step 4:appending data to existing table using append mode
#df1.write.format("delta").mode("append").saveAsTable("dev.demo_db.emp4")
#or
df1.write.format("delta").mode("overwrite").saveAsTable("dev.demo_db.emp4")
#here overwrite means-->oberwrites the existing data 
#append means-->appending new data to the xisting data
```

step 5: Retrieving data

```python
%sql
select * from dev.demo_db.emp4;
```

3rd method:  Dataframe insertInto() method

```python
#step 1: define schema 
from pyspark.sql.types import *
#Or
from pyspark.sql.types import StructType,StructField,IntegerType,StringType,DoubleType
schema2=StructType([
         StructField("eid",IntegerType(),False), \
         StructField("ename",StringType(),True), \
         StructField("sal",DoubleType(),True), \
         StructField("sex",StringType(),True), \
         StructField("dno",IntegerType(),True) \
         ])
```

```python
# step 2: create dataframe

data1=[(106,"Pavan",60000.00,"M",11),(107,"Suresh",70000.00,"M",12)]
#creating dataframe
df1=spark.createDataFrame(data=data1,schema=schema2)
display(df1)
```

```python
%sql
select * from dev.demo_db.emp4;
```

Insert the dataframe data into emp4 table

```python
# step 3: inserting---->using insertInto() method
#inserting data to existing table using insertInto
df1.write.insertInto("dev.demo_db.emp4",overwrite=False)

# now emp4 table should have 2 more records.
```

Retrieving data from emp4 table

```python
%sql
select * from dev.demo_db.emp4
```

```python
# 4th method : Insert using Temp view

df1.createOrReplaceTempView("delta_temp_table")
```

```python
%sql
select * from delta_temp_table
```

Insert delta_temp_table data into emp4 table.

```python
%sql
insert into dev.demo_db.emp4
select * from delta_temp_table
```

```python
%sql
select * from dev.demo_db.emp4
```

```python
# 5th method : spark sql insert
spark.sql("insert into dev.demo_db.emp4 select * from delta_temp_table")
```
