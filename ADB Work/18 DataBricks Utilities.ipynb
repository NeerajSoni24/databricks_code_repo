dbutils:

- dbutils is a package which provides many utilities.

- to display all the avialable utilities

dbutils.help()
- This module provides various utilities for users to interact with the rest of Databricks.


```python
dbutils.help()
```

1) Credentials: DatabricksCredentialUtils ->
- Utilities for interacting with credentials within notebooks

2) Data utility: 
To understand the datasets
- It provides a method to calculate some statistics and 
                gives you the summary
      
- DataUtils -> Utilities for understanding and interacting with datasets

3) fs: File System utility
- provides methods to interact with filesystem
- DbfsUtils -> Manipulates the Databricks filesystem (DBFS) from the console

4) Jobs Utility: To interact with databricks workflow jobs
- JobsUtils -> Utilities for leveraging jobs features

5) Library utility :
- LibraryUtils ---> allows you to locally compile an application

6) Meta utility
  MetaUtils ---> not for metadata purpose
- provides methods for a compiler

7) Notebook utility: To run chain of notebooks
- NotebookUtils -> Utilities for the control flow of a notebook 

8) Preview : 
for hidden utilities under preview

9) Secrets utility : for accessing secure credentials from within the notebook
- SecretUtils -> Provides utilities for leveraging secrets within notebooks

10) Widgets utility:
- WidgetsUtils --> To create notebook parameters


We can see the documentation of all these utilities using 2nd level help

ex: list of methods for fs utility


- dbutils.fs.help()  ----->displays all the methods for fs utility

- we can also check documentation for a specific command

dbutils.fs.help("cp")

In this way-->we can get help for any databricks utility

most commonly used utilities:
1. File Sytem utility: It provides methods to perform dbfs operations
                       such as cp,ls,head,mkdir,mv,rmdir etc

ex:ls

we can perform ls in 2 ways

1. using magic command (%fs ls /databricks-datasets)

2. using dbutils (dbutils.fs.ls("/databricks-datasets"))

1) using magic command %fs and ls 

```python
%fs
ls /databricks-datasets
```

2. using dbutils (dbutils.fs.ls)

```python
# in the above cell i didnt use the dbutils package

# 2) doing the same using--->dbutils package

dbutils.fs.ls("/databricks-datasets")

# The advantage of dbutils is--->
# dbutils.fs can be used in python code
# ex:
# for foldername in dbutils.fs.ls("/databricks-datasets"):
#  print(foldername.path)
```
