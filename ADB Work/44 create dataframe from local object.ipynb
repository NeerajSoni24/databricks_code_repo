Dataframe is a collection of Row objects,
each row object repreents a record

DataFrames can be created in 2 ways using

1)spark session->ex:df=spark.createDataFrame(RDD) #here spark is spark session object

2)sqlContext---->ex:df=sqlContext.createDataFrame(RDD)


Creating Dataframes :4ways

1)from local objects(python objects)

2)from rdds(resilient distributed datasets)

3)from the results of sql queries we get a DataFrame

4)from External Datasources like CSV,json etc


Creating Dataframe from local objects(python objects)

6)Creating a DF from a RDD

```python

x=[('Miller',25),('Ajay',30),('Blake',45)]
#converting a python object to spark object(RDD) using-->parallelize() method
rdd=sc.parallelize(x) #If we parallelize a python object we get a spark object(RDD)
#Now creating a spark dataframe from a spark RDD
df1=spark.createDataFrame(rdd,['Name','Age']) #providing schema while Creating DF
df1.show()
```

7)Creating a DataFrame from a Row object

```python
#creating a DF from a row object
x=[('Miller',25),('Ajay',30),('Blake',45)]
rdd=sc.parallelize(x)

#Now providing schema to a RDD using Row object
from pyspark.sql import Row
customer=Row('Name','age')
type(customer)
cust=rdd.map(lambda p:customer(*p))
cust.collect()  #returns a list of Row objects
#Now creating  DF from the Row objects
df=spark.createDataFrame(cust)
df.show()
```

8)Providing Schema using StructType

```python
from pyspark.sql.types import *
schema=StructType([
      StructField('Name',StringType(),True),
      StructField('age',IntegerType(),True)])
df=spark.createDataFrame(rdd,schema)
df.show()
```
