```python
# Reading data from Azure Sql database by the Azure databricks

# sql server name: mydbserver2021.database.windows.net
```

Go to Azure SQL database.
mine is: azsqldb2021 (mydbserver2021/azsqldb2021)

- Go to Overview
- then see server name and paste it under hostname.
- go to query editor and get database, user name and paste below.

```python
jdbcHostname="mydbserver2021.database.windows.net"
jdbcPort=1433
jdbcDatabase="azsqldb2021"
jdbcUsername="mydbuser2021"
jdbcPassword="Neeraj@2021"
jdbcDriver="com.microsoft.sqlserver.jdbc.SQLServerDriver"

jdbcUrl=f"jdbc:sqlserver://{jdbcHostname}:{jdbcPort};databaseName={jdbcDatabase};user={jdbcUsername};password={jdbcPassword}"
```

```python
# Now reading data(table) from azure sql and creating a df

df1=spark.read.format("jdbc").option("url",jdbcUrl).option("dbtable", "dbo.emp").load()
display(df1)
```

Another method
By using Connection String

Go to Azure SQL database
- go to overview then connection string
- select JDBC and copy the details from there and paste in below cell.
- provide the pwd { }

```python
connectionString="jdbc:sqlserver://mydbserver2021.database.windows.net:1433;database=azsqldb2021;user=mydbuser2021@mydbserver2021;password={Neeraj@2021};encrypt=true;trustServerCertificate=false;hostNameInCertificate=*.database.windows.net;loginTimeout=30;"
```

```python
# extracting data from azure sql by using connection string.
df=spark.read.jdbc(connectionString,"dbo.emp")
display(df)
```
