CRUD Operations

```python
# Step 1: Create Delta Table
from delta.tables import *

DeltaTable.create(spark) \
     .tableName("dev.demo_db.emp_4") \
     .addColumn("eid","INT") \
     .addColumn("ename","STRING") \
     .addColumn("sal","DOUBLE") \
     .addColumn("sex","STRING") \
    .addColumn("dno","INT") \
    .execute()
```

```python
%sql
select * from dev.demo_db.emp_4
```

Step 2: inserting data(sql style)

```python
%sql
insert into dev.demo_db.emp_4 values 
(101,"Raj",50000,"M",11),
(102,"Ravi",60000,"M",12),
(103,"Suresh",70000,"M",13);
```

Step 3: Retrieving data
1. sql style

```python
%sql
select * from dev.demo_db.emp_4
```

```python
# 2)pyspark style
# i) using show() method
spark.sql("select * from dev.demo_db.emp_4").show()
```

```python
# 1) display method
spark.read.format("delta").table("dev.demo_db.emp_4").display()
```

```python
# 2) using display() method another way
spark.sql("select * from dev.demo_db.emp_4").display()
```

```python
# 3) using display() function
display(spark.sql("select * from dev.demo_db.emp_4"))
```
