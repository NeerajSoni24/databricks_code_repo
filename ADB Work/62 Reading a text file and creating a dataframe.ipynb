```python
df=spark.read.text("/mnt/files/data/emp4.txt")
#or df=spark.read.format("text").load("/mnt/myfiles/data/emp4.txt")
df.show()
```

Text file doesnt contain header.

use selectExpr to split the values and create columns

```python
df1=df.selectExpr("split(value, ',')[0] as eid","split(value, ',')[1] as ename",
 "split(value, ',')[2] as sal", "split(value, ',')[3] as sex", 
"split(value, ',')[4] as dno")

df1.show()
```

Now we can see the column values in the df.

Another Method of Reading text file data with schema

First create schema

```python
schema1="""eid int,
           ename string,
           sal double,
           sex string,
           dno int"""
```

Now read text file data by using the schema1

```python
df_text=spark.read.schema(schema1).csv("/mnt/files/data/emp4.txt")
```

Now see the df_text dataframe data.

```python
df_text.show()
```
