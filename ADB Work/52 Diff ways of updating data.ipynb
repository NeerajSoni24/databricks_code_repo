2nd Method : Pyspark style using Table instance

```python
from delta.tables import *
from pyspark.sql.functions import *

deltatable_instance = DeltaTable.forName(spark,"dev.demo_db.emp4")
deltatable_instance.update(condition="ename='Ravi'",set={"sal":"90000"})
```

```python
%sql
select * from dev.demo_db.emp4;
```

```python
# ex:2 Declare the predicate by using spark sql functions

deltatable_instance.update(condition=col("ename") == "Ravi",set={"sal":lit("95000")})
```

```python
%sql
select * from dev.demo_db.emp4;
```
