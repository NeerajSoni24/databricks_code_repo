Pyspark Date and Timestamp functions

Datetype----->default format is---->

yyyy-mm-dd

Timestamptype->default format is---> 

yyyy-MM-dd HH:mm:ss.ssss

- Date and Time functions are very important if you are using pyspark for ETL.

- Pyspark SQL provides several Date and Timestamp functions.

```python
#from pyspark.sql import SparkSession
#from pyspark.sql.functions import *
#Create SparkSession
#spark = SparkSession.builder \
#        .appName("Spark DataFrames") \
#        .getOrCreate()
data=[["1","2022-03-01"],["2","2023-03-01"],["3","2024-03-01"]]
df=spark.createDataFrame(data,["id","date"])
df.show()
```


Pyspark Date Functions:

1. current_date(): Returns the current system date

```python
from pyspark.sql.functions import *
df.select(current_date().alias("current_date")).show(1)
```

```python
df.show()
```

Pyspark Date Functions:

2. date_format(): 

Converts date from yyyy-dd-mm to mm-dd-yyyy format

```python
df.select(col("date"),
           date_format(col("date"),"MM-dd-yyyy").alias("date_format")).show()
```

Pyspark Date Functions:

3. to_date(): 

converts string in date format yyyy-MM-dd to a datetype yyyy-MM-dd

```python
df1=df.select(col("date"),
          to_date(col("date"),"yyyy-MM-dd").alias("to_date"))
df1.show()
```

```python
df.printSchema()
df1.printSchema()
```

Pyspark Date Functions:

4. datediff(): returns the differnece between 2 dates

```python
df.select(col("date"),
          datediff(current_date(),col("date")).alias("datediff/No of days")).show()
```

5. months_between():

returns the months between 2 dates

```python
df.select(col("date"),
          months_between(current_date(),col("date")).alias("months_between")).show()
```

6. add_months() , date_add(),date_sub()

  adding  and substracting date and month from a given input

```python
df.select(col("date"),
          add_months(col("date"),3).alias("add_months"),
          add_months(col("date"),-3).alias("sub_months"),
          date_add(col("date"),4).alias("date_add"),
          date_sub(col("date"),4).alias("date_sub")).show()
```

7. year(), month(), next_day(), weekofyear()

```python
df.select(col("date"),
          year(col("date")).alias("year"),
          month(col("date")).alias("month"),
          next_day(col("date"),"Sunday").alias("next_day"),
          weekofyear(col("date")).alias("weekofyear")).show()
```

8. dayofweek(), dayofmonth(), dayofyear()

```python
df.select(col("date"),
          dayofweek(col("date")).alias("dayofweek"),
          dayofmonth(col("date")).alias("dayofmonth"),
          dayofyear(col("date")).alias("dayofyear")).show()
```

9. hour(), minute() and second()

```python
# hour(), minute and seconds
data=data=[["1","2022-03-01 11:01:19.07"],["2","2023-03-01 12:10:18.50"],["3","2024-03-01 10:25:30.01"]]
df2=spark.createDataFrame(data,["id","date"])
df2.show()
display(df2)
```
