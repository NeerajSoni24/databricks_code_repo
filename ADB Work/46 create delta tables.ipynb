1. SQL Approach

1)using create stmt--->creating table in dev catalog-->demo_db database

```python
%sql
create table if not exists dev.demo_db.emp1(
  eid int,
  ename string,
  sal int,
  sex string,
  dno int
) using delta;
```

we said--->using delta;

similarly 
for json--->say

using json;-->means store this data of table in json format

similarly-->for parquet
using parquet;

if it is a delta table or parquet table or json table

in databricks,default file format is delta,
so this will be a delta table

The data created for this table will be stored in delta table format

using delta means --->To make sure that it is a delta table

default also it is delta, if you won't specify any

Now Insert the values into the delta table

```python
%sql
insert into dev.demo_db.emp1 values (101, 'John',50000, 'M', 11);
```

```python
%sql
insert into dev.demo_db.emp1 values (102,'Jane',60000,'F',12),(103,'Mike',70000,'M',13),(104,'Sara',80000,'F',14);
```

```python
%sql
select * from dev.demo_db.emp1;
```

Another way of creating Delta table (DataFrame API).

Loading data into delta table (using DataFrame API-) -Another way of creating a Delta table

```python
emps_df.write.format("delta").mode("append").saveAsTable("dev.demo_db.emp")

# if table already exists it loads and saves into table
# If table doesnt exist-->It creates table and saves data into it
# schema is taken from dataframe
```

```python
%sql
select * from dev.demo_db.emp;
```

2. Second method of creating delta table --->
by pyspark approach

```python
#creating catalog ,database and table and loading data into it
spark.sql("create catalog if not exists dev")
spark.sql("create database if not exists dev.demo_db")

schema1="""eid int,
           ename string,
           sal int,
           sex string,
           dno int"""
# Read data from json file located at container
empdf=spark.read.format("json").schema(schema1).load("/mnt/files/jsondata/emps.json")

# Save the data as Delta table in dev.demo_db
empdf.write.format("delta").mode("append").saveAsTable("dev.demo_db.emp2")
```

```python
%sql
select * from dev.demo_db.emp2;
```

using DeltatableBuilder API


```python
from delta import DeltaTable  #or from delta.tables import *
(DeltaTable.create(spark) 
#(or)DeltaTable.createOrReplace(spark) #if already exists replace and create
#(or)DeltaTable.createIfNotExists(spark) #if not exists only-->create
                                         #here we wot see spaces unlike sql
           .tableName("dev.demo_db.emp_3")
           .addColumn("eid", "int")
           .addColumn("ename", "string")
           .addColumn("sal","int")
           .addColumn("sex","string")
           .addColumn("dno","int")
           .execute()
)
```
