```python
#syntax for join:
#----------------
#df1.join(df2,joining condition,"Type of Join")
#Types:
#1.inner
#2.outer
#3.left_outer
#4.right_outer
#5.full_outer
emps2_df = spark.read.json("/mnt/files/jsondata/emps2.json")
emps2_df.show()
dept_df = spark.read.json("/mnt/files/jsondata/dept.json")
dept_df.show()
```

```python
# 1)inner join
ij=emps2_df.join(dept_df,emps2_df.dno==dept_df.dno,"inner").select(emps2_df.eid,emps2_df.ename,emps2_df.sal,emps2_df.sex,emps2_df.dno,dept_df.dname,dept_df.city)
ij.show()
```

```python
# 2)left_outer:
loj=emps2_df.join(dept_df,emps2_df.dno==dept_df.dno,"left_outer").select(emps2_df.eid,emps2_df.ename,emps2_df.sal,emps2_df.sex,emps2_df.dno,dept_df.dname,dept_df.city)
loj.show()
```

```python
# 3)right_outer
roj=emps2_df.join(dept_df,emps2_df.dno==dept_df.dno,"right_outer").select(emps2_df.eid,emps2_df.ename,emps2_df.sal,emps2_df.sex,dept_df.dno,dept_df.dname,dept_df.city)
roj.show()
```

```python
# 4)full_outer:
foj=emps2_df.join(dept_df,emps2_df.dno==dept_df.dno,"full_outer").select(emps2_df.eid,emps2_df.ename,emps2_df.sal,emps2_df.sex,dept_df.dno,dept_df.dname,dept_df.city)
foj.show()
```
