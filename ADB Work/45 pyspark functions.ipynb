Pyspark functions:

1)withColumn(): we can perform the following

           1)for changing the column datatype

           2)for modifying or updating the column value
           
           3)Deriving a new column from the existing column
           
           4)Renaming a column
           
           5)dropping a dataframe column


```python
data=[(101,'Miller',50000,'2020-02-21','M','pune'),
      (102,'Blake',60000,'2019-04-15','M','hyd'),
      (103,'Sony',70000,'2021-06-25','F','pune'),
      (104,'alia',80000,'2018-03-17','F','hyd')
      ]
columns=["Empid","Empname","Salary","JoinDate","Sex","City"]
df=spark.createDataFrame(data=data,schema=columns)
df.show()
```

1) for changing the column datatype

  for changing th column datatype,we use cast()function along with withColumn()

df.printSchema()

o/p:

root

 |-- Empid: long (nullable = true)

 |-- Empname: string (nullable = true)

 |-- Salary: long (nullable = true)

 |-- JoinDate: string (nullable = true)

 |-- Sex: string (nullable = true)

 |-- City: string (nullable = true)

changing the datatype of Salary column  from long to int


```python
df.printSchema()
```

changing the datatype of Salary column from long to integer

```python
from pyspark.sql.functions import col
df1=df.withColumn("Salary",col("Salary").cast("Integer"))
df1.show()
```

```python
df1.printSchema()
```

2) for modifying or updating the column value
    
    i)updating with a value

   ii)updating based on condition

Task: Incrementing the salaries with a hike of 20%

```python
df2=df.withColumn("Salary",df.Salary+df.Salary*0.20)
df2.show()
```

Updating with a constant value

Task: adding 5000 as bonus to the sals of each employee

```python
df2=df.withColumn("Salary",df.Salary+5000)
df2.show()
```

ii) updating based on condition using when()

   -here we use withColumn() along with when()
   
       modify---->"M"------>"MALE"
               "F"------>"FEMALE"

```python
from pyspark.sql.functions import when
df3=df.withColumn("Sex",when(df.Sex=="M","MALE",)
                       .when(df.Sex=="F","FEMALE")
                       .otherwise(df.Sex))
df3.show()
```

3) Adding new column:
   -Adding a new column with default/constant/None/Null value
   -Adding a new column based on another column
   -Adding a new column based on condition

i)Adding a new column with constant

  -here we use a function called--->lit() function

Task: adding a new column (hike_percent)

Note: If you want to add a NULL/None-->then use-->lit(None)

```python
from pyspark.sql.functions import lit
df4=df.withColumn("Hike_percent",lit(0.30))
df4.show()
```

ii) Adding a new column based on another column

   Generate 2 new columns---> Tax and netsal 

   based on existing column-->Salary

```python
df5=df.withColumn("Tax",df.Salary*0.10)
df5=df5.withColumn("NetSalary",df5.Salary-df5.Tax)
df5.show()
```

------------------
iv) Generating a new column based on condition

   ex:Generate column Grade

                  if 
                  sal>=70000----------------->Grade "A"
                  
                  if sal>=50000 and sal<70000--->Grade "B"
                  
                  else                       --->Grade "C"


```python
from pyspark.sql.functions import when,lit
df.withColumn('GRADE',
          when((df.Salary>=70000),lit("A"))
          .when((df.Salary<70000) & (df.Salary>=50000),lit("B"))
          .otherwise(lit("C"))).show()
```

lit():

lit() function is used to add a constant value  as a new column

i) lit() function with select()

```python
from pyspark.sql.functions import col,lit
df2=df.select(col("Empid"),col("Empname"),col("Salary"),lit("IBM").alias("Company"))
df2.show()
```

ii) lit() function with withColumn()

```python
from pyspark.sql.functions import lit,col,when
df3=df.withColumn("Loan_status",when(col("Salary")>60000,lit("Eligible")) \
    .otherwise(lit("Not Eligible")))
df3.show()
```

withColumnRenamed():

Distinct() : For eliminating the duplicates

```python
data=[("Rahul","mrkt",30000),("Blake","sales",40000),("John","hr",50000),("Blake","sales",40000),("Rahul","mrkt",30000)]
columns=["Empname","Dept","Salary"]

df1=spark.createDataFrame(data=data,schema=columns)
df1.show()
```

```python
df2=df1.distinct()
df2.show()
```

I want the count of employees

```python
df2.show()
print("No of Employees=",df2.count())
```

dropDuplicates(): we can also remove the duplicates using this

```python
df1.show()
df3=df1.dropDuplicates()
df3.show()
```
